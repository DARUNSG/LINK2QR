# 01. Exploratory Data Analysis (EDA) & Data Leakage Discovery

This notebook explores the ISOT Fake News Dataset to analyze:
1. **Class Balance**: Distribution of Real vs Fake news labels.
2. **Text Length Distributions**: Word counts and character counts per class.
3. **Top Terms & N-grams**: Frequent terms distinguishing real vs fake news.
4. **Data Leakage Check**: Identifying agency tags like `(Reuters)` and city headers that act as artificial shortcut features.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Set plot styling
plt.style.use('ggplot')
plt.rcParams['figure.figsize'] = (10, 5)

# Import project data utilities
import sys
sys.path.append('..')
from src.data.download import generate_synthetic_data
from src.data.preprocess import load_and_preprocess_isot, detect_data_leakage
from src.utils.config import RAW_DATA_DIR

# Ensure raw data exists (generate synthetic if raw missing)
if not (RAW_DATA_DIR / 'True.csv').exists():
    generate_synthetic_data(num_samples=400)

df_true = pd.read_csv(RAW_DATA_DIR / 'True.csv')
df_fake = pd.read_csv(RAW_DATA_DIR / 'Fake.csv')
print(f'True news count: {len(df_true)}, Fake news count: {len(df_fake)}')
```

## 1. Class Balance Analysis

```python
df_true['label'] = 'Real'
df_fake['label'] = 'Fake'
combined_df = pd.concat([df_true, df_fake], ignore_index=True)

ax = sns.countplot(data=combined_df, x='label', palette=['#2ecc71', '#e74c3c'])
plt.title('Class Balance: Real vs Fake News')
plt.xlabel('Label')
plt.ylabel('Article Count')
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height() / 2),
                ha='center', va='center', fontsize=12, color='white', weight='bold')
plt.show()
```

## 2. Text Length Distribution

```python
combined_df['word_count'] = combined_df['text'].fillna('').apply(lambda x: len(x.split()))

sns.kdeplot(data=combined_df, x='word_count', hue='label', fill=True, common_norm=False, palette=['#2ecc71', '#e74c3c'])
plt.title('Word Count Distribution by Class')
plt.xlabel('Word Count')
plt.ylabel('Density')
plt.xlim(0, 1500)
plt.show()

combined_df.groupby('label')['word_count'].describe()
```

## 3. Data Leakage Pattern Discovery

In the ISOT dataset, real news articles consistently contain publisher attributions like `WASHINGTON (Reuters) -` or `(Reuters)`, whereas fake news articles lack these publisher tags. Models can easily achieve near 100% test accuracy by simply learning to look for the word 'Reuters', without actually understanding news content or credibility.

```python
df_raw = load_and_preprocess_isot(remove_leakage=False)
stats_uncleaned = detect_data_leakage(df_raw)

df_clean = load_and_preprocess_isot(remove_leakage=True)
stats_cleaned = detect_data_leakage(df_clean)

print('Data Leakage Summary:')
    print(f'Uncleaned Reuters Mention in Real News: {stats_uncleaned["reuters_mention_real_pct"]}%')
print(f'Uncleaned Reuters Mention in Fake News: {stats_uncleaned["reuters_mention_fake_pct"]}%')
```
